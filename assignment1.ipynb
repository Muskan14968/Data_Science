Q1 Banking: Take customer name and account balance, and print a welcome message with balance.

```python
name = input("Enter customer name: ")
account = float(input("Enter account balance: "))
print(f"Hi {name} Your account balance is: {account:.2f}")
```

Q2 E-commerce: Take product name and price, apply 10% discount and print final price.

```python
pro = input("Enter product name: ")
price = float(input("Enter price: "))
fprice = price * 0.9
print(f"Final price for {pro} after discount: {fprice:.2f}")
```
