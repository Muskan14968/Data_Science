Q1. Create a program that asks the user for 5 numbers and prints the maximum and minimum using loops.

```python
lit=[];
for i in range(5):
    num=int(input("Enter a number: "))
    lit.append(num)
max=lit[0];
min=lit[0];
for i in range(5):
    if lit[i]>max:
        max=lit[i]
    if lit[i]<min:
        min=lit[i]
print(f"Maximum number is {max} and Minimum number is {min}")
```

Q14. Write a program to check if a number is palindrome.

```python
num=int(input("Enter a number: "))
n=num
rev=0
while(num>0):
  rem=num%10
  rev=rev*10+rem
  num//=10
if rev==n:
  print("Palindrome")
else:
  print("Number is not palindrome")
```

Q20. Scenario: Password strength checker – check if password contains uppercase, lowercase, digit and special character.

```python
password=input("Enter your password: ")
l=0
u=0
spea=0
d=0
for i in password:
  if(i.islower()):
    l+=1
  elif(i.isupper()):
    u+=1
  elif(i.isdigit()):
    d+=1
  else:
    spea+=1
if(l>=1 and u>=1 and d>=1 and spea>=1):
  print("Your password is strong")
else:
  print("Your password is weak")
```

1. ATM Machine System (Loops + If-Else)
A user wants to use an ATM machine.

Requirements:
• User has 3 attempts to enter correct PIN.
• If correct → show menu:
o 1: Check Balance
o 2: Withdraw Money
o 3: Exit
• If withdraw amount > balance → show error.
• After each transaction, ask if user wants another transaction.

Concepts: while loop, if-else, break

```python
balance=10000
correct_pin=1520
attempts=3
while attempts>0:
  pin=int(input("Enter your pin: "))
  if pin==correct_pin:
    print("Login successful")
    while(True):
      print("\n1. Check Balance")
      print("2. Withdraw Money")
      print("3. Exit")
      ch = int(input("Enter your choice: "))
      if ch == 1:
        print(f"Your balance is {balance}")
      elif ch == 2:
        withdraw = int(input("Enter amount to withdraw: "))
        if withdraw > balance:
                    print("Insufficient balance")
        else:
                    balance -= withdraw
                    print("Transaction successful")
                    print(f"Remaining balance is {balance}")
      elif ch == 3:
                print("Thank you for using ATM")
                break
      else:
                print("Invalid choice")
    break
  else:
      attempts-=1
      print("Login failed")
if attempts==0:
        print("Account locked")
```

4. Electricity Bill Generator (IF-ELSE ONLY ❗)
Electricity bill is calculated based on units consumed. Requirements:

• Input number of units
• Charges:
o First 100 units → ₹5/unit
o Next 100 units → ₹7/unit
o Above 200 → ₹10/unit
• Add fixed charge ₹100
• Print total bill

Constraint: Use only if-else (no loops allowed)

```python
units=int(input("Enter no. of units"))
total=0
if units<=100:
  total+=units*5
  units-=100
if units<=100:
  total+=units*7
  units-=100
if units>200:
  total+=units*10
  units-=200
total+=100
print(f"Total bill: {total}")
```

• If strong → print "Password Created Successfully"
